# Implementación de una cola de tamaño fijo

```python
class Queue:
    def __init__(self, max_size=None):
        self.items = []
        self.max_size = max_size

    def is_empty(self):
        return len(self.items) == 0

    def is_full(self):
        if self.max_size is not None:
            return len(self.items) >= self.max_size
        return False

    def enqueue(self, item):
        if not self.is_full():
            self.items.append(item)
        else:
            print("La cola está llena. No se puede agregar el elemento:", item)

    def dequeue(self):
        if not self.is_empty():
            return self.items.pop(0)
        else:
            print("La cola está vacía")
            return None

    def size(self):
        return len(self.items)

    def print_front(self):
        if not self.is_empty():
            print("Primer elemento de la cola:", self.items[0])
        else:
            print("La cola está vacía")

    def print_all(self):
        if not self.is_empty():
            print("Elementos de la cola:")
            for item in self.items:
                print(item)
        else:
            print("La cola está vacía")
```

# creacion de una cola

```python
# Ejemplo de uso con un tamaño máximo de la cola
q = Queue(6)
```

```python
q.enqueue(1)
q.enqueue(2)
q.enqueue(3)
q.enqueue(54)
q.print_all()
```

Prueba del dequeue

```python
q.dequeue()
q.print_all()
```

```python
print("Tamaño de la cola:", q.size())
```

```python
class Nodo:
    def __init__(self, valor):
        self.valor = valor
        self.siguiente = None
class ColaDinamica:
    def __init__(self):
        self.frente = None
        self.final = None
        self.size = 0
        
    def is_empty(self):
        return self.frente is None
    
    def encolar(self, dato):
        new_nodo = Nodo(dato)
        if self.is_empty():
            self.frente = self.final = new_nodo
        else:
            self.final.siguiente = new_nodo
            self.final = new_nodo
        self.size += 1
            
    def desencolar(self):
        if self.is_empty():
            print("La cola está vacía")
            return None
        valor = self.frente.valor
        self.frente = self.frente.siguiente
        if self.frente is None:
            self.final = None
        self.size -= 1
        return valor
    
    def imprimir_todos(self):
        if self.is_empty():
            print("La cola está vacía")
            return
        actual = self.frente
        while actual:
            print(actual.valor)
            actual = actual.siguiente
        print()
    
    
```

```python
cola = ColaDinamica()
cola.encolar("Elemento 1")
cola.encolar("Elemento 2")
cola.encolar("Elemento 3")
cola.imprimir_todos()

print("elimiandon elemento 1",cola.desencolar())
cola.imprimir_todos()
```

Ejercicio: Simulación de una cola de atención al cliente

Crea un programa en Python que simule una cola de atención al cliente en una tienda. La cola debe permitir las siguientes operaciones:

    Agregar un cliente a la cola (opción 1).
    Atender al cliente al frente de la cola (opción 2).
    Mostrar el estado actual de la cola (opción 3).
    Salir del programa (opción 4).

Requisitos:

    Implementa la cola utilizando una lista en Python.
    Define las operaciones básicas de una cola:
        Encolar (enqueue).
        Desencolar (dequeue).
        Mostrar el estado de la cola (display).
    Usa un bucle para que el usuario pueda interactuar con el programa hasta que decida salir.

```python
class Cola:
    def __init__(self):
        self.cola=[]
        
    def enqueue(self, cliente):
        self.cola.append(cliente)
        print("Cliente", cliente, "agregado a la cola")
        
    def dequeue(self):
        if self.is_empty():
            print("la cola esta vacia, no hay clientes para atender")
        else:
            cliente_atendido = self.cola.pop(0)
            print("Cliente", cliente_atendido, "atendido")
            
    def display(self):
        return len(self.cola)
    
    def is_empty(self):
        return len(self.cola)==0
    
    
def main():
        cola = Cola()
        while True:
            print("Menunde cola de atencion de clientes")
            print("1. Agregar cliente")
            print("2. Atender cliente")
            print("3. Mostrar numero de clientes en cola")
            print("4. Salir")
            
            opcion = int(input("Ingrese una opcion: "))
            
            if opcion == 1:
                cliente = input("Ingrese el nombre del cliente: ")
                cola.enqueue(cliente)
            elif opcion == 2:
                cola.dequeue()
            elif opcion == 3:
                print("Numero de clientes en cola:", cola.display())
            elif opcion == 4:
                print("Gracias por usar nuestro servicio")
                break
            else: 
                print("Opcion invalida, intente de nuevo")
                
if __name__ == "__main__":
        main()
    
    
```
