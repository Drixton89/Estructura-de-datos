# Definición de una pila de tamaño fijo a partir de arreglos

```python
class Stack:
    def __init__(self, max_size=None):
        self.items = []
        self.max_size = max_size
        self.size = 0

    def is_empty(self):
        return len(self.items) == 0

    def is_full(self):
        return self.max_size is not None and self.size >= self.max_size

    def push(self, item):
        if self.is_full():
            raise OverflowError("La pila está llena (overflow) porfavor elimine un elemento")
        self.items.insert(0, item)
        self.size += 1

    def pop(self):
        if not self.is_empty():
            self.size -= 1
            return self.items.pop(0)
        else:
            raise IndexError("La pila está vacía (underflow)")

    def print_stack(self):
        print(self.items)

    def vaciar_pila(self):
        self.items = []
        self.size = 0

    def cima(self):
        if not self.is_empty():
            return self.items[0]
        else:
            raise IndexError("La pila está vacía (underflow)")
        
    def tamano(self):
        return self.size
```

```python
mi_pila = Stack(3)
```

```python
mi_pila.push(1)
mi_pila.push(2)
mi_pila.push(3)
mi_pila.print_stack()

```

```python
elemento_en_cima = mi_pila.cima()
print("Elemento en la cima:", elemento_en_cima)
```

```python
elemento_desapilado = mi_pila.pop()
print("Elemento desapilado:", elemento_desapilado)
mi_pila.print_stack()
```

```python
mi_pila.vaciar_pila()
mi_pila.print_stack()
```

#Definición de una pila de tamaño dinámico

```python
class DynamicStack:
    def __init__(self):
        self.items = []
        self.size = 0
        
    def is_empty(self):
        return len(self.items) == 0

    
    def push(self, item):
        self.items.insert(0, item)
        self.size += 1

    def pop(self):
        if not self.is_empty():
            self.size -= 1
            return self.items.pop(0)
        else:
            raise IndexError("La pila está vacía (underflow)")

    def print_stack(self):
        print(self.items)

    def vaciar_pila(self):
        self.items = []
        self.size = 0

    def cima(self):
        if not self.is_empty():
            return self.items[0]
        else:
            raise IndexError("La pila está vacía (underflow)")
        
    def tamano(self):
        return self.size
```

convierta un numero entero decimal a su representacion binaria utilizando una pila. no se permite el metodo bin()

```python
def decimal_a_binario(num):   #50
    pila = []
    
    while num > 0:
        resto = num % 2
        pila.append(resto)
        num = num // 2
        
    # 50 % 2 = 0, 50 // 2 = 25
    # 25 % 2 = 1, 25 // 2 = 12
    # 12 % 2 = 0, 12 // 2 = 6
    # 6 % 2 = 0, 6 // 2 = 3     
    # 3 % 2 = 1, 3 // 2 = 1
    # 1 % 2 = 1, 1 // 2 = 0
    # 0 % 2 = 0, 0 // 2 = 0
        
    binario = ''
    while pila:
        binario += str(pila.pop())
        
    return binario
        
    
```

```python
num = 50
resultado = decimal_a_binario(num)  
print(f"El número {num} en binario es: {resultado}")
```

```python
def navegador_web():
    historial = DynamicStack()

    # Simular visitas
    historial.push("www.google.com")
    historial.push("www.youtube.com")
    historial.push("www.wikipedia.org")

    print("Historial actual:")
    historial.print_stack()

    print("\nRetrocediendo...")
    ultima = historial.pop()
    print(f"Saliste de: {ultima}")
    print(f"Ahora estás en: {historial.cima()}")

    print("\nHistorial restante:")
    historial.print_stack()

# Ejecución
navegador_web()
```

```python
class Nodo:
    def __init__(self, dato):
        self.dato = dato
        self.siguiente = None

class PilaEnlazada:
    def __init__(self):
        self.cima = None
        self.tamano = 0

    def is_empty(self):
        return self.cima is None

    def push(self, dato):
        nuevo_nodo = Nodo(dato)
        nuevo_nodo.siguiente = self.cima
        self.cima = nuevo_nodo
        self.tamano += 1

    def pop(self):
        if self.is_empty():
            raise IndexError("La pila está vacía (underflow)")
        dato = self.cima.dato
        self.cima = self.cima.siguiente
        self.tamano -= 1
        return dato

    def peek(self):
        if self.is_empty():
            raise IndexError("La pila está vacía (underflow)")
        return self.cima.dato

    def size(self):
        return self.tamano

    def mostrar_pila(self):
        actual = self.cima
        print("Pila de acciones:")
        while actual:
            print("→", actual.dato)
            actual = actual.siguiente
```

```python
def bloc_de_notas():
    pila = PilaEnlazada()

    # Simular acciones
    pila.push("Escribió: 'Hola'")
    pila.push("Agregó: 'Mundo'")
    pila.push("Borró: 'Mundo'")
    pila.push("Agregó: 'GPT'")

    pila.mostrar_pila()
    print("\nDeshaciendo última acción...")
    print("Acción deshecha:", pila.pop())

    print("\nAcción actual:", pila.peek())
    print("Total de acciones restantes:", pila.size())
    print("\nEstado actual del historial:")
    pila.mostrar_pila()

# Ejecución
bloc_de_notas()
```

```python
def registro_llamadas():
    llamadas = DynamicStack()

    llamadas.push("Llamada a Mamá")
    llamadas.push("Llamada a Jefe")
    llamadas.push("Llamada a Clínica")
    llamadas.push("Llamada a Banco")

    print("Últimas llamadas:")
    llamadas.print_stack()

    print("\nEliminando última llamada...")
    print("Eliminada:", llamadas.pop())

    print("Llamada anterior:", llamadas.cima())
    print("Total de llamadas almacenadas:", llamadas.tamano())

# Ejecución
registro_llamadas()
```
