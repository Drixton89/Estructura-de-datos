```python
class DynamicArray:
    def __init__(self):
        # Número de elementos almacenados
        self.size = 0
        # Capacidad del buffer asignado
        self.capacity = 0
        # Buffer subyacente (lista de Python usada como array)
        self.array = []

    def append(self, x):
        # 1. Si no hay espacio suficiente, redimensionar
        if self.size + 1 > self.capacity:
            old_cap = self.capacity

            # Cálculo de la nueva capacidad (estrategia de sobreasignación)
            if old_cap < 9:
                extra = 3
            else:
                extra = 6
            new_capacity = old_cap + (old_cap >> 3) + extra

            # Reservar el nuevo buffer y copiar elementos antiguos
            new_array = [None] * new_capacity
            for i in range(self.size):
                new_array[i] = self.array[i]

            # Reemplazar buffer y actualizar capacidad
            self.array = new_array
            self.capacity = new_capacity

        # 2. Insertar el nuevo elemento
        self.array[self.size] = x
        self.size += 1

    def __len__(self):
        return self.size

    def __getitem__(self, index):
        if not 0 <= index < self.size:
            raise IndexError('Index out of bounds')
        return self.array[index]

    def __repr__(self):
        return f'DynamicArray([{", ".join(repr(self.array[i]) for i in range(self.size))}])'
```

## Resumir los métodos de inserción en uno solo

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class LinkedList:
    def __init__(self):
        self.Head = None

    def printList(self):
        current = self.Head
        while current is not None:
            print(current.data, end=" ")
            current = current.next

    def insertAtBeginning(self, element):
        if self.Head is None:
            newNode = Node(element)
            self.Head = newNode
        else:
            newNode = Node(element)
            newNode.next = self.Head
            self.Head = newNode


    def insertAtEnd(self, element):
        if self.Head is None:
            newNode = Node(element)
            self.Head = newNode
        else:
            current = self.Head
            while current.next is not None:
                current = current.next
            newNode = Node(element)
            current.next = newNode


    def insertAtPosition(self, element, position):
        counter = 1
        current = self.Head
        while counter < position - 1 and current is not None:
            counter += 1
            current = current.next
        if position == 1:
            newNode = Node(element)
            newNode.next = current
            self.Head = newNode
        else:
            newNode = Node(element)
            newNode.next = current.next
            current.next = newNode

    def deleteFromBeginning(self):
        if self.Head is None:
            print("La lista enlazada está vacía, no se puede eliminar un elemento.")
            return
        else:
            node = self.Head
            self.Head = self.Head.next
            del node

    def deleteFromLast(self):
        if self.Head is None:
            print("La lista enlazada está vacía, no se puede eliminar un elemento.")
            return
        else:
            current = self.Head
            previous = None
            while current.next is not None:
                previous = current
                current = current.next
            if previous is not None:
               previous.next = None
            else:
               self.Head = None
            del current

    def deleteAtPosition(self, position):
        if self.Head is None:
            print("La lista enlazada está vacía, no se puede eliminar un elemento.")
            return
        else:
            current = self.Head
            previous = None
            count = 1
            while current.next is not None and count < position:
                previous = current
                current = current.next
                count += 1
            if current == self.Head:
                self.Head = current.next
                del current
            else:
                previous.next = current.next
                del current

    def countElements(self):
        count = 0
        current = self.Head
        while current is not None:
            count += 1
            current = current.next
        print("El numero de elementos en la lista es: ", count)

    def replaceElement(self, old_element, new_element):
        current = self.Head
        while current is not None:
            if current.data == old_element:
                current.data = new_element
                break
            current = current.next

    def updateAtPosition(self, new_element, position):
        if self.Head is None and position != 1:
            print("No element to update in the linked list.")
            return
        elif self.Head is None and position == 1:
            newNode = Node(new_element)
            self.Head = newNode
            return
        count = 1
        current = self.Head
        while current.next is not None and count < position:
            count += 1
            current = current.next
        if count == position:
            current.data = new_element
        elif current.next is None:
            print("No hay tantos elementos en la lista enlazada")

    def printListReverse(self):
        self._printListReverseHelper(self.Head)

    def _printListReverseHelper(self, node):
        if node is None:
            return
        self._printListReverseHelper(node.next)
        print(node.data, end=" ")
```
