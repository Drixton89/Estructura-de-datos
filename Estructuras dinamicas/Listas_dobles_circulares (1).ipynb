```python
#metodo contrsctor de la clase nodo para lista enlazada simple
class Nodo:
    def __init__(self, dato):
        self.dato = dato   # el valor o dato del nodo
        self.siguiente = None # puntero al siguiente nodo, inicial en None
        
```

```python
class ListaEnlazadaSimple:
    def __init__(self):
       self.head = None # puntero al primer nodo de la lista, inicial en None
       
    def inserter_final(self, dato):
        nuevo_nodo = Nodo(dato)
        if self.head is None:
            self.head = nuevo_nodo
        else:
            actual = self.head
            while actual.siguiente:
                actual = actual.siguiente
            actual.siguiente = nuevo_nodo
            
    def mostrar_lista(self):
        actual = self.head
        while actual is not None:
            print(actual.dato, end=' -> ')
            actual = actual.siguiente
        print('None')
        
        
    def imprimir_reverso(self, head):
        if head is None:
            return
        self.imprimir_reverso(head.siguiente)
        #print("None <-", head.dato, end=' ')
        print(head.dato, end=' -> ')
           
    def imprimir_reverso2(self):
        # Usar una lista temporal para almacenar los datos
        elementos = []
        actual = self.head
        # Recorrer la lista y almacenar los datos
        while actual:
            elementos.append(actual.dato)
            actual = actual.siguiente
        # Imprimir en orden inverso
        for dato in reversed(elementos):
            print(dato, end=" -> ")
        print("None")
        
        
    def imprimir_alreves(self): #guisao
        actual=self.head
        lista=[]      
        while actual:
            lista.append(actual.dato)
            actual=actual.siguiente
       
        while lista:
            print(lista.pop(), ' ', end='')
        print()
    
            
```

### Definición de la clase doubleList

```python
class doubleList:
  def __init__(self):
    self.head = None

  def insertar_lista_vacia(self,dato):

    if self.head is None:
      nuevoNodo = Nodo(dato)
      self.head = nuevoNodo
    else:
      print("La lista no está vacía")

  def insertar_siguiente(self,dato):

    if self.head is None:
      self.insertar_lista_vacia(dato)
      return

    else:
      # si la lista no está vacía buscamos el elemento final:
      apuntador = self.head

      while apuntador.siguiente is not None:
        apuntador = apuntador.siguiente

      nuevoNodo = Nodo(dato)
      apuntador.siguiente = nuevoNodo
      nuevoNodo.anterior = apuntador

  def insertar_inicio(self,dato):

    if self.head is None:
      self.insertar_lista_vacia(dato)
      return

    nuevoNodo = Nodo(dato)

    nuevoNodo.siguiente = self.head
    self.head.anterior = nuevoNodo
    self.head = nuevoNodo

  def insertar_despues(self, valor, nuevo_dato):
        nuevoNodo = Nodo(nuevo_dato)
        apuntador = self.head
        while apuntador:
            if apuntador.elemento == valor:
                nuevoNodo.siguiente = apuntador.siguiente
                if apuntador.siguiente:
                    apuntador.siguiente.anterior = nuevoNodo
                apuntador.siguiente = nuevoNodo
                nuevoNodo.anterior = apuntador
                return
            apuntador = apuntador.siguiente

  def insertar_antes(self, valor, nuevo_dato):
        nuevoNodo = Nodo(nuevo_dato)
        apuntador = self.head
        while apuntador:
            if apuntador.elemento == valor:
                nuevoNodo.anterior = apuntador.anterior
                if apuntador.anterior:
                    apuntador.anterior.siguiente = nuevoNodo
                else:
                    self.head = nuevoNodo
                nuevoNodo.siguiente = apuntador
                apuntador.anterior = nuevoNodo
                return
            apuntador = apuntador.siguiente


  def imprimir(self):
        apuntador = self.head
        while apuntador:
            print(apuntador.elemento, end=" <-> ")
            apuntador = apuntador.siguiente
        print("None")

  def imprimir_reverso(self):
        apuntador = None
        if self.head:
            apuntador = self.head
            while apuntador.siguiente:
                apuntador = apuntador.siguiente

        while apuntador:
            print(apuntador.elemento, end=" <-> ")
            apuntador = apuntador.anterior

  def eliminar(self, elemento):
        apuntador = self.head
        while apuntador:
            if apuntador.elemento == elemento:
                if apuntador.anterior:
                    apuntador.anterior.siguiente = apuntador.siguiente
                if apuntador.siguiente:
                    apuntador.siguiente.anterior = apuntador.anterior
                if apuntador == self.head:
                    self.head = apuntador.siguiente
                return
            apuntador = apuntador.siguiente
            
  def insertar_final(self, nuevo_dato):
    nuevoNodo = Nodo(nuevo_dato)
    apuntador = self.head
    if not apuntador:
      self.head = nuevoNodo
      return

    while apuntador.siguiente is not None:
      apuntador = apuntador.siguiente

    apuntador.siguiente = nuevoNodo
    nuevoNodo.anterior = apuntador
```

### El nodo se define igual que en una lisa enlazada simple

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
```

```python
class CircularLinkedList:
    def __init__(self):
        self.head = None

    def append(self, data):
        new_node = Node(data)
        if not self.head:
            self.head = new_node
            self.head.next = self.head
        else:
            temp = self.head
            while temp.next != self.head:
                temp = temp.next
            temp.next = new_node
            new_node.next = self.head

    def display(self):
        if not self.head:
            return
        temp = self.head
        while True:
            print(temp.data, end=' -> ')
            temp = temp.next
            if temp == self.head:
                break


    def count(self):
        count = 0
        temp = self.head
        if not temp:
            return count
        count = 1
        temp = temp.next
        while temp != self.head:
            count += 1
            temp = temp.next
        return count
```

Dado head, el inicio de una lista enlazada, determina si la lista enlazada tiene un ciclo en ella.

Hay un ciclo en una lista enlazada si existe algún nodo en la lista que puede ser alcanzado nuevamente al seguir continuamente el puntero next. Internamente, pos se utiliza para indicar el índice del nodo al cual el puntero next del último nodo está conectado. Nota que pos no se pasa como parámetro.

Devuelve true si hay un ciclo en la lista enlazada. De lo contrario, devuelve false.

Ejemplo 1:

Entrada: head = [3,2,0,-4], pos = 1
Salida: true
Explicación: Hay un ciclo en la lista enlazada, donde la cola se conecta al primer nodo (indexado desde 0).

Ejemplo 2:

Entrada: head = [1,2], pos = 0
Salida: true
Explicación: Hay un ciclo en la lista enlazada, donde la cola se conecta al nodo 0.

Ejemplo 3:

Entrada: head = [1], pos = -1
Salida: false
Explicación: No hay un ciclo en la lista enlazada.

```python
from typing import Optional
class ListNode:
    def __init__(self, val=0, next=None):
        self.val=val
        self.next = next

        
class Solution:
    def ciclo(self, head) -> bool:
        s = f = head
        while f and f.next:
            s =s.next        #avanzo nodo
            f = f.next.next # avanzo dos nodos
            if s == f:
                return True
            
        return False
        
def crear_lista(elemetos, pos):
    head = ListNode(elemetos[0]) if elemetos else None
    actual = head
    node_circular = None
    
    for  index, e in enumerate(elemetos[:]):
        actual.next = ListNode(e)
        actual = actual.next
        if index == pos:
            node_circular = actual
            
    if pos != -1 and node_circular:
        actual.next = node_circular
        
    return head 
```

```python
class Node:
    def __init__(self, dato =0, siguiente = None):
        self.dato = dato
        self.siguiente = siguiente
        
def detectar_ciclo(head):
    if not head or not head.siguiente:
        return None
    #detectar ciclo en la lista 
    puntero_1 = head
    puntero_2 = head
    
    while puntero_2 and puntero_2.siguiente:
        puntero_1 = puntero_1.siguiente
        puntero_2 = puntero_2.siguiente.siguiente
        if puntero_1 == puntero_2:
            break
        else:
            return None
        
    # buscar el inicio del ciclo
    puntero_1 = head
    while puntero_1 != puntero_2:
        puntero_1 = puntero_1.siguiente
        puntero_2 = puntero_2.siguiente
    return puntero_2
        
```

Dadas las cabezas de dos listas enlazadas simples headA y headB, devuelve el nodo en el que las dos listas se intersectan. Si las dos listas enlazadas no tienen ninguna intersección, devuelve none.

Por ejemplo, las siguientes dos listas enlazadas comienzan a intersectarse en el nodo c1:

Los casos de prueba se generan de manera que no hay ciclos en ninguna parte de toda la estructura enlazada.

Nota que las listas enlazadas deben conservar su estructura original después de que la función retorne.

Juez Personalizado:

Las entradas al juez se dan de la siguiente manera (tu programa no recibe estas entradas):

intersectVal - El valor del nodo donde ocurre la intersección. Este es 0 si no hay un nodo de intersección.
listA - La primera lista enlazada.
listB - La segunda lista enlazada.
skipA - El número de nodos que se deben saltar en listA (comenzando desde la cabeza) para llegar al nodo de intersección.
skipB - El número de nodos que se deben saltar en listB (comenzando desde la cabeza) para llegar al nodo de intersección.
El juez luego creará la estructura enlazada basada en estas entradas y pasará las dos cabezas, headA y headB, a tu programa. Si devuelves correctamente el nodo de intersección, entonces tu solución será aceptada.

```python
from typing import Optional

class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

class Solution:
    def getIntersectionNode(self, headA: ListNode, headB: ListNode) -> Optional[ListNode]:
        if not headA or not headB:
            return None
        
        # Usar dos punteros para caminar a través de las dos listas
        pointerA, pointerB = headA, headB

        # Si uno de ellos llega al final, comenzar en la cabeza del otro
        while pointerA != pointerB:
            pointerA = pointerA.next if pointerA else headB
            pointerB = pointerB.next if pointerB else headA
        
        return pointerA  # Este será el nodo de intersección o None si no intersectan

# Funciones auxiliares para crear listas enlazadas
def create_linked_list(elements):
    if not elements:
        return None
    head = ListNode(elements[0])
    current = head
    for value in elements[1:]:
        current.next = ListNode(value)
        current = current.next
    return head

# Ejemplos de prueba
sol = Solution()

# Creación de nodos comunes
common = create_linked_list([8, 4, 5])

# Ejemplo 1
headA1 = create_linked_list([4, 1])
headA1.next.next = common  # Conexión a nodo común
headB1 = create_linked_list([5, 6, 1])
headB1.next.next.next = common  # Conexión a nodo común
print("Ejemplo 1 Salida:", sol.getIntersectionNode(headA1, headB1).val if sol.getIntersectionNode(headA1, headB1) else "No intersection")

# Ejemplo 2
headA2 = create_linked_list([1, 9, 1, 2, 4])
headB2 = create_linked_list([3])
headB2.next = headA2.next.next.next  # Conexión a nodo común
print("Ejemplo 2 Salida:", sol.getIntersectionNode(headA2, headB2).val if sol.getIntersectionNode(headA2, headB2) else "No intersection")

# Ejemplo 3
headA3 = create_linked_list([2, 6, 4])
headB3 = create_linked_list([1, 5])
print("Ejemplo 3 Salida:", sol.getIntersectionNode(headA3, headB3) if sol.getIntersectionNode(headA3, headB3) else "No intersection")
```

Dada una lista enlazada simple, devuelve el valor de un nodo aleatorio de la lista enlazada. Cada nodo debe tener la misma probabilidad de ser elegido.

Implementa la clase Solution:

Solution(ListNode head) Inicializa el objeto con la cabeza de la lista enlazada simple head.
int getRandom() Elige un nodo al azar de la lista y devuelve su valor. Todos los nodos de la lista deben tener la misma probabilidad de ser elegidos.
Ejemplo 1:

Entrada

[null, 1, 3, 2, 2, 3]

Explicación

solution.getRandom(); // devuelve 2

// getRandom() debe devolver 1, 2 o 3 aleatoriamente. Cada elemento debe tener igual probabilidad de devolver.

```python
import random

class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

class Solution:
    def __init__(self, head: Optional[ListNode]):
        self.head = head

    def getRandom(self) -> int:
        # Muestreo reservorio
        current = self.head
        chosen_value = 0
        i = 1
        while current:
            if random.randint(1, i) == 1:
                chosen_value = current.val
            current = current.next
            i += 1
        return chosen_value

# Funciones auxiliares para crear la lista enlazada
def create_linked_list(elements):
    head = ListNode(elements[0]) if elements else None
    current = head
    for value in elements[1:]:
        current.next = ListNode(value)
        current = current.next
    return head

# Caso de prueba
head = create_linked_list([1, 2, 3])
obj = Solution(head)
print(obj.getRandom())  # Devuelve aleatoriamente 1, 2 o 3
print(obj.getRandom())  # Devuelve aleatoriamente 1, 2 o 3
print(obj.getRandom())  # Devuelve aleatoriamente 1, 2 o 3
```

```python
class Node:
    def __init__(self, key, value):
        self.key = key
        self.value = value
        self.prev = None
        self.next = None

class LRUCache:

    def __init__(self, capacity: int):
        self.capacity = capacity
        self.cache = {}
        self.head = Node(0, 0)  # Dummy head
        self.tail = Node(0, 0)  # Dummy tail
        self.head.next = self.tail
        self.tail.prev = self.head

    def _remove(self, node):
        prev = node.prev
        next = node.next
        prev.next = next
        next.prev = prev

    def _add(self, node):
        p = self.tail.prev
        p.next = node
        node.prev = p
        node.next = self.tail
        self.tail.prev = node

    def get(self, key: int) -> int:
        if key in self.cache:
            node = self.cache[key]
            self._remove(node)
            self._add(node)
            return node.value
        return -1

    def put(self, key: int, value: int) -> None:
        if key in self.cache:
            self._remove(self.cache[key])
        node = Node(key, value)
        self.cache[key] = node
        self._add(node)
        if len(self.cache) > self.capacity:
            # Remove the LRU from the list
            lru = self.head.next
            self._remove(lru)
            del self.cache[lru.key]
```

Se te proporciona la cabeza de una lista enlazada no vacía que representa un número entero no negativo sin ceros a la izquierda.

Devuelve la cabeza de la lista enlazada después de duplicar su valor.

Por ejemplo:

Entrada: head = [1,8,9]

Salida: [3,7,8]

Explicación: La lista enlazada dada representa el número 189. Por lo tanto, la lista enlazada devuelta representa el número 189 * 2 = 378.

Entrada: head = [9,9,9]

Salida: [1,9,9,8]

Explicación: La lista enlazada dada representa el número 999. Por lo tanto, la lista
 enlazada devuelta representa el número 999 * 2 = 1998.

```python

from typing import Optional
class ListNode:
    def __init__(self, value=0, next=None):
        self.value = value
        self.next = next

class Solution:
    def doubleIt(self, head: Optional[ListNode]) -> Optional[ListNode]:
        if not head:
            return None
        
          
        
        # Convertir la lista enlazada a entero
        num = 0
        current = head
        while current:
            num = num * 10 + current.value
            current = current.next
        
        # Duplicar el número y convertir de nuevo a lista enlazada
        doubled_num = num * 2
        dummy_head = ListNode(0)
        current = dummy_head
        
        # Añadir cada dígito del número duplicado a la lista enlazada
        for char in str(doubled_num):
            current.next = ListNode(int(char))
            current = current.next
        
        return dummy_head.next


```

```python
def print_list(node):
    while node:
        print(node.value, end=' ')
        node = node.next
    print()
```

```python
# Crear instancias de ListNode para cada prueba
node1 = ListNode(1, ListNode(2, ListNode(3)))
node2 = ListNode(5)
node3 = ListNode(9, ListNode(9, ListNode(9)))
node4 = ListNode(1, ListNode(0, ListNode(0, ListNode(1))))

# Crear instancia de Solution
solution = Solution()

# Realizar pruebas
result1 = solution.doubleIt(node1)
result2 = solution.doubleIt(node2)
result3 = solution.doubleIt(node3)
result4 = solution.doubleIt(node4)

# Imprimir resultados
print("Test 1: ", end="")
print_list(result1)  # Esperado: 2 4 6
print("Test 2: ", end="")
print_list(result2)  # Esperado: 1 0
print("Test 3: ", end="")
print_list(result3)  # Esperado: 1 9 9 8
print("Test 4: ", end="")
print_list(result4)  # Esperado: 2 0 0 2
```

Una empresa quiere simular la disposición de empleados alrededor de una mesa redonda para un juego de rotación de turnos. Se debe implementar una lista doblemente enlazada circular donde:

    Se puedan agregar personas al final.

    Se pueda mostrar el orden de los participantes hacia adelante y hacia atrás.

    Se pueda eliminar una persona por nombre.

    Se pueda mover n turnos adelante o atrás desde una persona específica.

```python
class NodoCircular:
    def __init__(self,nombre):
        self.nombre = nombre
        self.siguiente = None
        self.anterior = None
        
class ListaCircular:
    def __init__(self):
        self.head = None
        
    def  insertar(self, nombre):
        new_nodo = NodoCircular(nombre)
        if self.head is None:
            new_nodo.siguiente = new_nodo.anterior = new_nodo
            self.head = new_nodo
        else:
            last = self.head.anterior
            new_nodo.siguiente = self.head
            new_nodo.anterior = last
            last.siguiente = new_nodo
            self.head.anterior = new_nodo
            
            
    def mostrar_adelante(self, cantidad):
        if self.head is None:
            print("La lista está vacía")
            return
        actual = self.head
        for _ in range(cantidad):
            print(actual.nombre, end=" -> ")
            actual = actual.siguiente
        print("...")
        
    def mostrar_atras(self, cantidad):
        if not self.head:
            print("La lista está vacía")
            return
        actual = self.head.anterior
        for _ in range(cantidad):
            print(actual.nombre, end=" -> ")
            actual = actual.anterior
        print("...")
        
    def eliminar(self, nombre):
        if self.head is None:
            print("La lista está vacía")
            return
        actual = self.head
        while True:
            if actual.nombre == nombre:
                if actual.siguiente == actual:          
                    self.head = None
                else:
                    actual.anterior.siguiente = actual.siguiente
                    actual.siguiente.anterior = actual.anterior
                    if actual == self.head:
                        self.head = actual.siguiente
                print(f"Eliminado: {nombre}")
                return
            actual = actual.siguiente
            if actual == self.head:
                break
        print(f"{nombre} no encontrado en la lista")
                    
                    
            
                
        
            
        
   
        
        
```
