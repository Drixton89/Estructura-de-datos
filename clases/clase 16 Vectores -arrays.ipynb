# Broadcasting (Difusion)

Los arreglos de Numpy difieren de una lista normal de Python en su capacidad de Broadcasting, qeu es asignar un valor a un rango de posiciones:

```python
# definiendo un valor para todo un rango de posiciones (Broadcasting)
arr[1:5]=100 # Asignar el numero 100 a las pocisiones desde el 0 hasta el 4
arr
```

```python
#crear nuevamente el arreglo con el que estabamos trabajando
arr = np.arange(0,11)
arr
```

```python
#NOTA importante en la seleccion de rangos (sclicing)
# los arreglos son mutables
slice_of_arr = arr[2:6]
slice_of_arr
```

```python
#Cambiar todos los valores a 99
slice_of_arr[:]=99
slice_of_arr
```

Observe que los cambios tambien ocurren en el arreglo original

```python
arr
```

Los datos no se copian, ¡es un puntero a el arreglo original! ¡Esto evita problemas de memoria!

```python
# Para obtener una copia, se debe hacer explicito
arr = np.arange(0,11)
arr_copy = arr.copy()
arr_copy[:] = 99
arr_copy
```

```python
# Observe que el arreglo original no se modifico
arr
```

Ejercicio: Operaciones con Vectores

    Crear dos vectores aleatorios
        Usa NumPy para crear dos vectores de tamaño 5, cada uno con valores aleatorios entre 0 y 10.

    Realizar operaciones básicas
        Calcula la suma, resta, multiplicación y división de los dos vectores creados.

    Calcular la magnitud
        Calcula la magnitud (norma) de cada vector.

    Producto escalar
        Calcula el producto escalar de los dos vectores.

    Ángulo entre los vectores
        Calcula el ángulo (en grados) entre los dos vectores.

```python
import numpy as np

vector_a = np.random.randint(0,10,5)
vector_b = np.random.randint(0,10,5)
print("Vector A: ",vector_a)
print("Vector B: ",vector_b)
```

```python
suma = vector_a + vector_b
resta = vector_a - vector_b
multiplicacion = vector_a * vector_b
#division = vector_a / vector_b
div =np.divide(vector_a,vector_b, out=np.zeros_like(vector_a, dtype=float), where=vector_b!=0)

print("Suma: ",suma)
print("Resta: ",resta)
print("Multiplicacion: ",multiplicacion)
#print("Division: ",division)
print("Division pro: ",div)
```

```python
magnitud_a = np.linalg.norm(vector_a)
magnitud_b = np.linalg.norm(vector_b)
print("Magnitud de A: ",magnitud_a)
print("Magnitud de B: ",magnitud_b)
```

```python
productor_escalar = np.dot(vector_a,vector_b) #producto punto
print("Producto escalar: ",productor_escalar)
pp=vector_a@vector_b
print("Producto escalar: ",pp)
```

```python
if magnitud_a * magnitud_b != 0:
    cos_angulo = productor_escalar/(magnitud_a*magnitud_b)
else:
    cos_angulo = 0
    
angulo =np.degrees(np.arccos(np.clip(cos_angulo, -1.0, 1.0)))

print("angulo entre A y B: ",angulo,"°")
```

Nota: Una matriz es simétrica si es igual a su transpuesta.

```python
def es_simetrica():
    matriz=np.random.randint(0,11,(3,3))
    es_simetrica = (matriz == matriz.T).all()
    print("Matriz: \n",matriz)
    print("Es simetrica: ",es_simetrica)
    simetrica= np.array_equal(matriz, matriz.T)
    print("Es simetrica: ",simetrica)
    
es_simetrica()
```

Suma de los elementos de la diagonal principal y secundaria

Descripción:
Crea una matriz cuadrada de tamaño 4×4 con números enteros aleatorios entre 1 y 20. Calcula la suma de los elementos de su diagonal principal y su diagonal secundaria.

```python
def sumnadiag():
    matriz=np.random.randint(1,21,(4,4))
    suma_diagonal_ppal = np.trace(matriz)
    suma_diagonal_sec = np.trace(matriz[:,::-1])
    diag_sec= np.fliplr(matriz).diagonal().sum()
    diag_ppal= matriz.diagonal().sum()
    diagsec2=np.trace(np.fliplr(matriz))
    print("Matriz: \n",matriz)
    print("Suma diagonal principal: ",suma_diagonal_ppal)
    print("Suma diagonal secundaria: ",suma_diagonal_sec)
    print("Suma diagonal principal: ",diag_ppal)
    print("Suma diagonal secundaria: ",diag_sec)
    print("Suma diagonal secundaria: ",diagsec2)
    
    
sumnadiag()
    
    
```

Encontrar la fila con mayor suma

Descripción:
Crea una matriz de tamaño 5×3 con números enteros aleatorios entre 0 y 15. Calcula la suma de los elementos de cada fila e identifica cuál fila tiene la mayor suma. Muestra la fila y su suma.

```python
def mayor_suma():
    matriz= np.random.randint(0,16,(5,3))
    suma_filas = matriz.sum(axis=1)
    fila_max= np.argmax(suma_filas)
    print("Matriz: \n",matriz)
    print("Suma de las filas: ",suma_filas)
    print("Fila con mayor sumapor indice: ",fila_max)
    print(f"Fila con mayor suma : {fila_max+1} (suma: {suma_filas[fila_max]})")
    
mayor_suma()
```

Ejercicio 1: Calcular la cantidad de producto formado en una reacción química

Enunciado:
En la reacción 2H2 + 02 ​→ 2H20 , se tienen 5 moles de H2​ y 3 moles de O2​. Usando matrices, determina cuántos moles de H20 se pueden formar, y verifica si sobra alguno de los reactivos.

FORMULAS
CANTIDAD DE PRODUCTO FORMADO = MIN(REACTIVO DISPONIBLE/COEFICENTE ESTEQUIOMETRICO)* COEFICEINTE DE PRODUCTO



```python
import numpy as np
coeficeintes= np.array([2,1,2])
reactivos = np.array([5,3])

print(coeficeintes[2])

MF= min(reactivos/coeficeintes[:2])*coeficeintes[2]
print("Maximo de producto final: ",MF)
sobrantes = reactivos - (MF/ coeficeintes[2])*coeficeintes[:2]
print("Sobrantes: ",sobrantes)
```

Balance de cargas en una solución iónica

Enunciado:
En una solución se tienen los siguientes iones:

    Na+ con concentración 0.10M.
    Cl− con concentración 0.08M.
    SO24−​ con concentración 0.04M.

Determina si la solución es eléctricamente neutra.

Fórmula a usar:

    Carga total=∑(concentracioˊn del ion×valencia)
    Si Carga total=0 la solución es neutra.

```python
iones = np.array([
    [0.10,+1],
                 [0.08, -1],
                 [0.04, -2]
                 ])

cargatotal= np.sum(iones[:,0]*iones[:,1])
print("Carga total: ",cargatotal)
if cargatotal==0:
    print("La molecula es neutra")
```

Calcula el pH total de la solución, teniendo en cuenta la contribución de cada ácido.

formulas 
concentracion total de  H+ = sumatoria(concentracion de acifdo * numero de H+)
